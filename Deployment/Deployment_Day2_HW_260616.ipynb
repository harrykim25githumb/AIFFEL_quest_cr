1-1. FastAPI가 Flask보다 모델 배포에 적합한 이유 세 가지는 무엇입니까?

→ 비동기 처리 지원, 자동 API 문서 생성(Swagger), Pydantic 기반 데이터 검증 기능을 제공하기 때문입니다.

1-2. Uvicorn의 역할은 무엇이며, 왜 FastAPI와 함께 사용합니까?

→ Uvicorn은 FastAPI 애플리케이션을 실행하는 ASGI 서버이며, 비동기 처리를 효율적으로 지원하기 위해 사용합니다.

1-3. @app.get("/health")에서 get과 "/health"는 각각 무엇을 의미합니까?

→ get은 HTTP GET 메서드, "/health"는 요청 URL 경로를 의미합니다.

1-4. FastAPI에서 dict를 반환하면 어떤 일이 자동으로 일어납니까?

→ JSON으로 변환되어 HTTP 응답으로 반환됩니다.



2-1. /models/sentiment-v1에서 sentiment-v1은 어떤 종류의 파라미터입니까?

→ Path Parameter(경로 파라미터)입니다.

2-2. /models?status=running&limit=5에서 status와 limit은 어떤 종류의 파라미터입니까?

→ Query Parameter(쿼리 파라미터)입니다.

2-3. 모델 추론 요청에 Request Body를 사용하는 이유는 무엇입니까?

→ 입력 데이터가 크고 복잡하여 URL보다 Request Body로 전달하는 것이 적합하기 때문입니다.

2-4. FastAPI에서 함수의 파라미터가 Path, Query, Body 중 어디서 오는지 어떻게 판별합니까?

→ {} 안의 변수는 Path, 기본 타입 파라미터는 Query, Pydantic 모델은 Body로 판단합니다.



3-1. FastAPI에서 Swagger UI에 접속하려면 어떤 URL로 이동합니까?

→ 기본적으로 http://localhost:8000/docs 입니다.

3-2. Swagger UI가 코드와 항상 동기화될 수 있는 이유는 무엇입니까?

→ FastAPI가 코드로부터 OpenAPI 스키마를 자동 생성하기 때문입니다.

3-3. Pydantic 모델의 Field(description=, examples=)는 Swagger UI의 어디에 반영됩니까?

→ description은 필드 설명에, examples는 Example Value 영역에 표시됩니다.

3-4. Swagger UI와 ReDoc의 핵심 차이는 무엇입니까?

→ Swagger UI는 API 테스트가 가능하고, ReDoc은 문서 열람에 최적화되어 있습니다.



4-1. 모델을 서버 시작 시 한 번만 로드해야 하는 이유는 무엇입니까?

→ 매 요청마다 모델을 로드하면 성능이 저하되고 메모리를 비효율적으로 사용하기 때문입니다.

4-2. pixel_values가 784개가 아닌 요청이 들어오면 어떤 일이 발생합니까? 이를 처리하는 코드를 직접 작성했습니까?

→ 422 Validation Error가 발생하며, Pydantic이 자동으로 검증하므로 별도 코드를 작성하지 않았습니다.

4-3. HTTPException(status_code=503)은 어떤 상황에서 사용했습니까? 왜 500이 아니라 503입니까?

→ 모델이 아직 로드되지 않은 경우 사용했으며, 서버 오류가 아니라 일시적인 서비스 불가 상태이기 때문입니다.

4-4. Swagger UI에서 PredictRequest의 description과 examples가 어디에 표시됩니까?

→ Request Body 섹션의 필드 설명과 Example Value 영역에 표시됩니다.
