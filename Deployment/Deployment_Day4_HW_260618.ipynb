### 1-1. Streamlit의 스크립트 재실행 모델이란 무엇입니까?

사용자의 입력이 발생할 때마다 Streamlit 스크립트 전체를 처음부터 다시 실행하는 방식입니다.

### 1-2. `st.text_input()`에 값을 입력하면 내부적으로 어떤 일이 일어납니까?

입력값이 상태에 저장되고, 스크립트 전체가 다시 실행됩니다.

### 1-3. `st.set_page_config()`를 스크립트 중간에 호출하면 어떻게 됩니까?

오류가 발생합니다. 반드시 Streamlit 명령 중 가장 먼저 호출해야 합니다.

----------

### 2-1. `st.file_uploader()`로 업로드된 파일의 바이트 데이터는 어떻게 얻습니까?

`uploaded_file.getvalue()` 또는 `uploaded_file.read()`를 사용합니다.

### 2-2. Streamlit에서 `@st.cache_resource`를 사용하는 이유는 무엇입니까?

모델, DB 연결, API 클라이언트 같은 리소스를 한 번만 생성하고 재사용하기 위해 사용합니다.


----------

### 3-1. 모놀리식과 분리 아키텍처의 핵심 차이를 한 문장으로 설명하세요.

모놀리식은 UI와 모델 서버가 하나의 애플리케이션이고, 분리 아키텍처는 서로 독립된 서버로 운영됩니다.

### 3-2. 모델을 업데이트할 때, 분리 아키텍처에서는 어떤 서버만 재배포하면 됩니까?

FastAPI 모델 서버만 재배포하면 됩니다.

### 3-3. Streamlit 앱에 PyTorch가 설치되어 있지 않아도 되는 이유는 무엇입니까?

실제 추론은 FastAPI 서버에서 수행하고 Streamlit은 API 호출만 하기 때문입니다.

----------


### 4-1. 이미지를 API에 전송할 때 Base64로 인코딩하는 이유는 무엇입니까?

바이너리 이미지를 JSON으로 안전하게 전송할 수 있는 문자열 형태로 변환하기 위해서입니다.

### 4-2. `response.raise_for_status()`는 어떤 역할을 합니까?

HTTP 응답이 4xx 또는 5xx일 때 예외를 발생시켜 에러를 쉽게 처리할 수 있게 합니다.
