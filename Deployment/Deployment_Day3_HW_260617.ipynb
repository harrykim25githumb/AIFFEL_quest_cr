각 섹션별 체크포인트 답변 



1-1. time.sleep(3)과 await asyncio.sleep(3)의 핵심 차이는 무엇입니까?

    : time.sleep()은 현재 스레드를 멈추고, asyncio.sleep()은 대기 중에 다른 작업이 실행되도록 이벤트 루프를 양보합니다.

1-2. 모델 추론처럼 CPU를 계속 사용하는 작업에서 async/await만으로 동시 처리가 안 되는 이유는 무엇입니까?

    : async/await는 I/O 대기에 효과적이며, CPU를 계속 점유하는 작업은 이벤트 루프를 양보하지 않기 때문입니다.


2-1. async def 안에서 time.sleep(3)을 호출하면 왜 다른 요청까지 지연됩니까?

    : 이벤트 루프가 3초 동안 완전히 멈춰 다른 요청을 처리할 수 없기 때문입니다.

2-2. 일반 def로 선언된 엔드포인트는 FastAPI가 내부적으로 어떻게 처리합니까?

    : FastAPI가 자동으로 스레드풀(ThreadPool)에서 실행합니다.

2-3. blocking 추론 중 헬스체크까지 막히면 실무에서 어떤 문제가 발생할 수 있습니까?

    : 로드밸런서나 오케스트레이터가 서버를 장애 상태로 오판할 수 있습니다.


3-1. run_in_executor가 이벤트 루프 블로킹을 방지하는 원리는 무엇입니까?

    : CPU 작업을 별도 스레드나 프로세스로 넘겨 이벤트 루프가 계속 요청을 처리하도록 합니다.

3-2. run_in_executor의 첫 번째 인자에 None을 넣으면 어떤 스레드풀이 사용됩니까?

    : asyncio의 기본 ThreadPoolExecutor가 사용됩니다.

3-3. 일반 def와 async def + run_in_executor의 핵심 차이는 무엇입니까?

    : 일반 def는 전체 함수가 기본 스레드풀에서 실행되고, async def + run_in_executor는 필요한 부분만 스레드풀로 보낼 수 있으며 연산용 스레드풀을 별도 설정 가능합니다.

3-4. GPU 추론 시 스레드풀 크기를 1~2로 제한하는 이유는 무엇입니까?

    : 과도한 동시 GPU 접근으로 인한 메모리 부족과 성능 저하를 방지하기 위해서입니다.


4-1. 글로벌 Exception Handler를 사용하면 어떤 반복을 줄일 수 있습니까?

    : 각 엔드포인트마다 동일한 예외 처리 코드를 작성하는 반복을 줄일 수 있습니다.

4-2. 클라이언트에게 스택 트레이스를 노출하면 안 되는 이유는 무엇입니까?

    : 내부 구조와 민감한 정보가 노출되어 보안 위험이 발생할 수 있기 때문입니다.

4-3. logging 모듈이 print()보다 나은 점은 무엇입니까?

    : logging 모듈은 에러 발생 시 시간, 심각도, 모듈명 등 상세 정보를 제공합니다.
