1-1. Day 5(정형 데이터)와 Day 7(텍스트 생성)에서 전처리가 어떻게 다릅니까?
   : Day 5는 수치형 데이터를 정규화·인코딩하고, Day 7은 텍스트를 토큰 ID로 변환합니다.

1-2. Hugging Face Transformers 라이브러리의 역할은 무엇입니까?
  : 사전학습 언어모델과 토크나이저를 쉽게 로드하고 추론할 수 있게 해주는 라이브러리입니다.

---

2-1. 토크나이저의 `encode()`와 `decode()`는 각각 어떤 변환을 수행합니까?
  : `encode()`는 텍스트를 토큰 ID로 변환하고, `decode()`는 토큰 ID를 텍스트로 변환합니다.

2-2. `model.generate()`의 `temperature` 파라미터는 생성 결과에 어떤 영향을 줍니까?
  : 값이 높을수록 다양하고 창의적인 답변이 나오며, 낮을수록 일관적이고 보수적인 답변이 나옵니다.

2-3. 멀티턴 대화에서 이전 대화 기록을 프롬프트에 포함하는 이유는 무엇입니까?
  : 이전 대화 맥락(context)을 파악하여 자연스럽고 일관된 답변을 생성하기 위해서입니다.

---

3-1. `ChatRequest`의 `messages`가 리스트인 이유는 무엇입니까?
  : 여러 개의 대화 턴(사용자·챗봇 메시지)을 '순서대로' 전달하기 위해서입니다.

3-2. 서버가 대화 기록을 저장하지 않고, 클라이언트가 매번 전체 기록을 보내는 이유는?
  : 서버를 상태 없는(stateless) 구조로 유지하여 구현과 확장을 단순하게 하고 비용을 효율화 하기 위해서입니다.

3-3. `max_new_tokens`을 너무 크게 설정하면 어떤 문제가 발생할 수 있습니까?
  : 응답 시간이 길어지고 메모리 사용량이 증가할 수 있습니다.

---

4-1. `st.chat_message`와 `st.chat_input`은 각각 어떤 역할을 합니까?
  : `st.chat_message`는 대화를 표시하고, `st.chat_input`은 사용자 입력을 받습니다.

4-2. `st.session_state["chat_messages"]`에 대화 기록을 저장하는 이유는?
  : Streamlit 재실행 후에도 대화 내역을 유지하기 위해서입니다.

4-3. "대화 초기화" 버튼이 `st.rerun()`을 호출하는 이유는 무엇입니까?
  : 상태를 초기화한 뒤 화면을 즉시 새로고침하여 변경사항을 반영하기 위해서입니다.

---

5-1. 챗봇 API에서 인증이 실패하면 서버는 어떤 상태 코드를 반환합니까?
  : HTTP 401 (Unauthorized) 상태 코드를 반환합니다.

5-2. UI의 `call_chat_api()`는 401 응답을 받으면 사용자에게 무엇을 보여줍니까?
  : API Key가 없거나 유효하지 않다는 오류 메시지를 보여줍니다.

5-3. Day 6의 인증 모듈(`app/auth.py`)을 Day 7에서 수정 없이 재사용할 수 있는 이유는 무엇입니까?
  : 인증 로직이 모델 종류와 무관한 공통 기능으로 분리되어 있기 때문입니다.
