### 1-1. 인증 없는 API가 위험한 이유를 두 가지 이상 설명하세요.

누구나 API를 호출할 수 있어 무단 사용이 가능하며, 과도한 요청으로 서버 비용 증가나 서비스 장애가 발생할 수 있습니다.

### 1-2. API Key 방식이 ML 추론 API에 적합한 이유는 무엇입니까?

구현이 간단하고 서버 간 통신이나 추론 API 접근 제어에 충분한 보안을 제공합니다. 기계-기계 간 접속 방식에 적합.

---

### 2-1. `Header(None)`에서 `None`은 어떤 상황에서 `x_api_key`에 들어갑니까?

요청 헤더에 `X-API-Key`가 포함되지 않았을 때 `None`이 전달됩니다.

### 2-2. `Depends(verify_api_key)`를 엔드포인트에 추가하면 요청 처리 흐름이 어떻게 바뀝니까?

엔드포인트 실행 전에 `verify_api_key()`가 먼저 실행되어 인증을 검증합니다.

### 2-3. 인증 실패 시 반환하는 HTTP 상태 코드 401의 의미는 무엇입니까?

"인증실패 에러 401 Unauthorized" : 인증 정보가 없거나 유효하지 않아 접근이 거부되었음을 의미합니다.

---

### 4-1. `UploadFile`과 Base64 방식의 핵심 차이는 무엇입니까?

`UploadFile`은 파일을 직접 전송하고, Base64는 파일을 문자열로 변환하여 JSON에 포함해 전송합니다.

### 4-2. `file.content_type`으로 타입을 검증하는 이유는 무엇입니까?

이미지가 아닌 파일 업로드를 차단하여 오류와 보안 문제를 예방하기 위해서입니다. 가장 저렴한 검사를 가장 먼저 진행하여 효율화.

### 4-3. 파일 크기를 제한하지 않으면 어떤 문제가 발생할 수 있습니까?

메모리 부족, 서버 성능 저하, 서비스 거부(DoS) 공격 위험이 발생할 수 있습니다.

---

### 6-1. API Key 인증이 없으면 어떤 위험이 있습니까? (두 가지 이상)

서버 무단 사용, 서버 비용 증가, 과도한 요청에 의한 서비스 장애가 발생할 수 있습니다.

### 6-2. Depends(verify_api_key)는 엔드포인트 실행 전에 어떤 일을 합니까?

API Key를 검증하고 인증 실패 시 요청을 즉시 차단합니다.

### 6-3. UploadFile 방식이 Base64보다 편리한 점은 무엇입니까?

JSON 타입으로의 인코딩 과정이 필요 없고 전송 크기가 더 작아 효율적입니다.

### 6-4. 파일 업로드 시 크기 제한을 하지 않으면 어떤 문제가 생깁니까?

서버가 다운되거나 서버 메모리 사용량이 급증하고 서비스 성능이 저하될 수 있습니다.

### 6-5. 이미지를 28x28 그레이스케일로 변환하는 이유는 무엇입니까?

MNIST 모델이 학습한 입력 형식과 동일하게 맞추기 위해서입니다.
